```python

# ============================================================
# REGIONAL ENROLLMENT ANALYSIS
# World Bank Global Education Insights
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

# ------------------------------------------------------------
# 1. Validate the required datasets and columns
# ------------------------------------------------------------

required_selected = [
    'Country Code',
    'Indicator Name',
    'Year',
    'Value'
]

required_country = [
    'Country Code',
    'Region',
    'Income Group'
]

missing_selected = [
    col for col in required_selected
    if col not in selected_df.columns
]

missing_country = [
    col for col in required_country
    if col not in country_df.columns
]

if missing_selected:
    raise ValueError(
        f"Missing columns in selected_df: {missing_selected}"
    )

if missing_country:
    raise ValueError(
        f"Missing columns in country_df: {missing_country}"
    )

# ------------------------------------------------------------
# 2. Select enrollment observations
# ------------------------------------------------------------

enrollment_df = selected_df[
    selected_df['Indicator Name']
    .str.contains('enrolment', case=False, na=False)
].copy()

if enrollment_df.empty:
    raise ValueError(
        "No enrollment indicators found in selected_df. "
        "Check the indicator names in your dataset."
    )

enrollment_df['Year'] = pd.to_numeric(
    enrollment_df['Year'], errors='coerce'
)

enrollment_df['Value'] = pd.to_numeric(
    enrollment_df['Value'], errors='coerce'
)

enrollment_df = enrollment_df.dropna(
    subset=['Country Code', 'Year', 'Value']
)

# ------------------------------------------------------------
# 3. Attach World Bank regional and income-group metadata
# ------------------------------------------------------------

metadata_cols = [
    'Country Code',
    'Region',
    'Income Group'
]

metadata = country_df[metadata_cols].drop_duplicates(
    subset=['Country Code']
)

enrollment_df = enrollment_df.merge(
    metadata,
    on='Country Code',
    how='left',
    validate='many_to_one'
)

# Exclude regional aggregates and entities without a region
enrollment_df = enrollment_df.dropna(
    subset=['Region']
).copy()

print("Enrollment records:", len(enrollment_df))
print(
    "Unique reporting entities:",
    enrollment_df['Country Code'].nunique()
)
print("\nRegions available:")
print(sorted(enrollment_df['Region'].dropna().unique()))

print("\nEnrollment indicators available:")
print(
    enrollment_df['Indicator Name']
    .drop_duplicates()
    .to_list()
)

# ------------------------------------------------------------
# 4. Chart 1: Average enrollment by region and education level
# ------------------------------------------------------------

regional_summary = (
    enrollment_df
    .groupby(['Region', 'Indicator Name'], as_index=False)
    .agg(
        Average_Enrollment=('Value', 'mean'),
        Reporting_Entities=('Country Code', 'nunique')
    )
)

plt.figure(figsize=(13, 7))

sns.barplot(
    data=regional_summary,
    x='Region',
    y='Average_Enrollment',
    hue='Indicator Name'
)

plt.title(
    'Average Enrollment Rate by World Bank Region '
    'and Education Level'
)
plt.xlabel('World Bank Region')
plt.ylabel('Average Enrollment Rate (%)')
plt.xticks(rotation=30, ha='right')
plt.legend(
    title='Education Level',
    bbox_to_anchor=(1.02, 1),
    loc='upper left'
)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 5. Chart 2: Enrollment distribution by region
# ------------------------------------------------------------

plt.figure(figsize=(13, 7))

sns.boxplot(
    data=enrollment_df,
    x='Region',
    y='Value',
    hue='Indicator Name',
    showfliers=False
)

plt.title(
    'Distribution of Enrollment Rates Across Regions '
    'and Education Levels'
)
plt.xlabel('World Bank Region')
plt.ylabel('Enrollment Rate (%)')
plt.xticks(rotation=30, ha='right')
plt.legend(
    title='Education Level',
    bbox_to_anchor=(1.02, 1),
    loc='upper left'
)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 6. Chart 3: Regional enrollment trends over time
# ------------------------------------------------------------

regional_trends = (
    enrollment_df
    .groupby(
        ['Year', 'Region', 'Indicator Name'],
        as_index=False
    )
    .agg(
        Average_Enrollment=('Value', 'mean'),
        Reporting_Entities=('Country Code', 'nunique')
    )
)

# Keep years with at least 3 reporting entities per
# region-indicator combination to reduce very sparse averages.
regional_trends = regional_trends[
    regional_trends['Reporting_Entities'] >= 3
].copy()

regions = sorted(regional_trends['Region'].unique())

if regions:
    ncols = 2
    nrows = int(np.ceil(len(regions) / ncols))

    fig, axes = plt.subplots(
        nrows, ncols,
        figsize=(16, 5 * nrows),
        squeeze=False
    )

    for i, region in enumerate(regions):
        ax = axes[i // ncols, i % ncols]

        region_data = regional_trends[
            regional_trends['Region'] == region
        ]

        sns.lineplot(
            data=region_data,
            x='Year',
            y='Average_Enrollment',
            hue='Indicator Name',
            marker='o',
            ax=ax
        )

        ax.set_title(region)
        ax.set_xlabel('Year')
        ax.set_ylabel('Average Enrollment Rate (%)')
        ax.legend(fontsize=8)

    for j in range(len(regions), nrows * ncols):
        axes[j // ncols, j % ncols].set_visible(False)

    fig.suptitle(
        'Enrollment Trends by World Bank Region',
        fontsize=16,
        y=1.01
    )
    plt.tight_layout()
    plt.show()
else:
    print(
        "Not enough reporting entities for the regional "
        "trend chart. Review reporting coverage."
    )

# ------------------------------------------------------------
# 7. Chart 4: Regional enrollment by income group
# ------------------------------------------------------------

income_summary = (
    enrollment_df
    .dropna(subset=['Income Group'])
    .groupby(
        ['Income Group', 'Indicator Name'],
        as_index=False
    )
    .agg(
        Average_Enrollment=('Value', 'mean'),
        Reporting_Entities=('Country Code', 'nunique')
    )
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=income_summary,
    x='Income Group',
    y='Average_Enrollment',
    hue='Indicator Name'
)

plt.title(
    'Average Enrollment Rate by Income Group '
    'and Education Level'
)
plt.xlabel('World Bank Income Group')
plt.ylabel('Average Enrollment Rate (%)')
plt.xticks(rotation=25, ha='right')
plt.legend(
    title='Education Level',
    bbox_to_anchor=(1.02, 1),
    loc='upper left'
)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 8. Summary table for interpretation
# ------------------------------------------------------------

print("\nREGIONAL ENROLLMENT SUMMARY")
display(
    regional_summary.sort_values(
        ['Indicator Name', 'Average_Enrollment'],
        ascending=[True, False]
    ).round(2)
)

print("\nINCOME-GROUP ENROLLMENT SUMMARY")
display(
    income_summary.sort_values(
        ['Indicator Name', 'Average_Enrollment'],
        ascending=[True, False]
    ).round(2)
)

print("\nREGIONAL REPORTING COVERAGE")
display(
    regional_trends.groupby(
        ['Region', 'Indicator Name']
    ).agg(
        First_Year=('Year', 'min'),
        Last_Year=('Year', 'max'),
        Number_of_Years=('Year', 'nunique'),
        Minimum_Reporting_Entities=(
            'Reporting_Entities', 'min'
        )
    ).reset_index()
)

# ------------------------------------------------------------
# 9. Interpretation guide
# ------------------------------------------------------------

print("""
HOW TO INTERPRET THESE RESULTS

1. Regional averages:
   Identify which regions report higher or lower average
   enrollment at each education level.

2. Boxplots:
   Examine differences between countries within each region.
   A wide box or long whiskers indicate greater variation.

3. Regional trends:
   Examine whether average enrollment changes over time.
   Reporting coverage may vary by year and region.

4. Income groups:
   Compare enrollment averages across World Bank income groups.

IMPORTANT LIMITATIONS

- These are descriptive comparisons, not causal findings.
- Country observations are not weighted by population.
- The number and mix of reporting countries can differ.
- Regional averages do not directly measure student retention.
- Interpret results only after reviewing the reporting
  coverage and the actual values in the summary tables.
""")
```

### Regional Enrollment Analysis: Key Findings

**Objective:** To examine differences in average enrollment rates across World Bank regions and education levels.

**Key findings**

1. **Differences across education levels:** Primary enrollment averages are higher than lower-secondary and upper-secondary enrollment averages in all seven regions. This highlights the importance of examining participation beyond primary education.

2. **Regional disparities:** Sub-Saharan Africa has the lowest average enrollment rates across primary (73.68%), lower-secondary (24.41%), and upper-secondary education (13.21%) in the available observations.

3. **Higher reported enrollment in Europe & Central Asia:** The region records average enrollment rates of 96.89% at the primary level, 88.95% at the lower-secondary level, and 76.57% at the upper-secondary level.

4. **Variation within regions:** The boxplots show that enrollment rates vary across countries within the same region. Regional averages alone therefore do not capture the full range of country-level performance.

5. **Income-group differences:** The income-group summary also shows lower average enrollment at higher education levels among low-income groups. These patterns warrant further examination using consistent country coverage and appropriate education indicators.

**Business impact**

The findings can help education policymakers and development organizations identify regions and education levels that may require greater attention. The results support further investigation into access barriers, transition pathways, school availability, and participation in secondary education.

**Analytical limitations**

These results are descriptive and based on unweighted averages of available observations. Reporting coverage differs by region, indicator, and year.
```python
# Merge Region and Income Group

adult_region_df = adult_country_df.merge(
    country_df[['Country Code', 'Region', 'Income Group']],
    on='Country Code',
    how='left'
)

print(adult_region_df.shape)

adult_region_df.head()
```

```python
# Check regions available

adult_region_df['Region'].value_counts()
```

```python
adult_region_df.shape
```

```python
# Average Adult Literacy Rate by Region

region_literacy = adult_region_df.groupby('Region')['Value'] \
                                 .mean() \
                                 .sort_values(ascending=False)

print(region_literacy)
```

```python
plt.figure(figsize=(10,6))

region_literacy.plot(
    kind='bar',
    color='steelblue'
)

plt.title('Average Adult Literacy Rate by Region (2011)')
plt.xlabel('Region')
plt.ylabel('Adult Literacy Rate (%)')

plt.xticks(rotation=30, ha='right')

plt.show()
```

**Observation from Regional Literacy Analysis**

The regional analysis reveals significant disparities in adult literacy rates across the world. Europe & Central Asia recorded the highest average adult literacy rate (98.2%), indicating near-universal literacy among adults. Latin America & Caribbean (92.5%) and East Asia & Pacific (92.0%) also demonstrated strong literacy outcomes.

In contrast, South Asia recorded the lowest average adult literacy rate (52.5%), followed by Sub-Saharan Africa (72.4%). These results highlight persistent educational challenges and unequal access to learning opportunities across regions.

**Business Impact**

The findings suggest that literacy improvement initiatives should be prioritized in South Asia and Sub-Saharan Africa, where adult literacy levels remain substantially below global leaders. Governments and development organizations may focus on adult education programs, community-based learning initiatives, and targeted interventions for marginalized populations. High-performing regions can serve as benchmarks for effective education policies and literacy promotion strategies.

```python
# Average Adult Literacy by Income Group

income_literacy = adult_region_df.groupby('Income Group')['Value'] \
                                 .mean() \
                                 .sort_values(ascending=False)

print(income_literacy)
```

```python
plt.figure(figsize=(10,6))

income_literacy.plot(
    kind='bar',
    color='green'
)

plt.title('Average Adult Literacy Rate by Income Group (2011)')
plt.xlabel('Income Group')
plt.ylabel('Adult Literacy Rate (%)')

plt.xticks(rotation=20)

plt.show()
```

**Observation from Income Group Analysis**

The analysis demonstrates a strong positive relationship between national income levels and adult literacy rates. High-income countries, both OECD and non-OECD, recorded literacy rates above 97%, indicating near-universal literacy among adults.

Upper-middle-income countries also exhibited strong literacy performance (94.5%). However, literacy rates declined substantially among lower-middle-income countries (79.3%) and low-income countries (52.7%).

The gap of nearly 45 percentage points between high-income and low-income countries highlights the significant influence of economic development on educational attainment and literacy outcomes.

**Business Impact**

The results suggest that economic development and educational achievement are closely interconnected. Countries with greater financial resources are better positioned to invest in schools, teacher training, educational infrastructure, and adult learning programs.

Development agencies and policymakers should prioritize literacy improvement programs in low-income and lower-middle-income countries, where literacy deficits remain substantial and may constrain economic growth, workforce productivity, and social development.

## Conclusion

This exploratory data analysis examined global education indicators from the World Bank EdStats dataset, with a focus on adult literacy, youth literacy, and enrollment across primary, lower-secondary, and upper-secondary education.

The literacy analysis explored trends over time, differences between adult and youth literacy, the distribution of adult literacy observations, and variations across countries, regions, and income groups. The country-level rankings and regional comparisons provide a descriptive view of differences in educational outcomes among the entities reporting data.

The enrollment analysis examined changes over time and compared average enrollment rates across education levels, World Bank regions, and income groups. The results show that average primary enrollment is higher than average enrollment at lower-secondary and upper-secondary levels across the regions examined. The regional summaries also show differences in reported enrollment levels, highlighting areas that warrant further investigation.

These findings should be interpreted in the context of data availability and coverage. The number of countries reporting information varies across indicators and years, and regional averages are not necessarily based on the same countries throughout the period. The analysis is descriptive and does not establish causal relationships. Differences between enrollment rates at different education levels should not be interpreted as direct measures of student dropout or retention.

Overall, the analysis demonstrates the value of exploratory data analysis for identifying patterns, comparing educational indicators, and highlighting areas for further research. Future analysis could examine gender-specific education outcomes, education expenditure, tertiary participation, and more direct indicators of student progression and completion, subject to the availability and quality of the relevant data.
